```python
# -*- coding: cp1251 -*-
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

case = 'a'          # 'a' или 'b'
h = 0.1           # шаг интегрирования
t_end = 20          # время моделирования

# Параметры ДУ
if case == 'a':
    m, p, k = 1, 1, 1          
else:
    m, p, k = 1, 0.75, 0.125   

# Начальные условия
x0, v0 = 1.0, 0.0
y0 = [x0, v0]

# Матрица системы
A = np.array([[0, 1], [-k/m, -p/m]])

# ТОЧНОЕ РЕШЕНИЕ
def exact_solution(t):
    if case == 'a':
        sqrt3 = np.sqrt(3)
        return np.exp(-t/2) * (np.cos(sqrt3*t/2) + (1/sqrt3)*np.sin(sqrt3*t/2))
    else:
        return 2*np.exp(-t/4) - np.exp(-t/2)

# ЯВНЫЙ ЭЙЛЕР
def explicit_euler():
    n = int(t_end/h) + 1
    t = np.linspace(0, t_end, n)
    y = np.zeros((n, 2))
    y[0] = y0
    for i in range(n-1):
        y[i+1] = y[i] + h * (A @ y[i])
    return t, y[:, 0]

# НЕЯВНЫЙ ЭЙЛЕР
def implicit_euler():
    n = int(t_end/h) + 1
    t = np.linspace(0, t_end, n)
    y = np.zeros((n, 2))
    y[0] = y0
    I = np.eye(2)
    for i in range(n-1):
        # Решаем (I - hA)y_{n+1} = y_n
        y[i+1] = np.linalg.solve(I - h*A, y[i])
    return t, y[:, 0]

# ПОЛУЯВНЫЙ ЭЙЛЕР 
def semi_euler():
    n = int(t_end/h) + 1
    t = np.linspace(0, t_end, n)
    y = np.zeros((n, 2))
    y[0] = y0
    I = np.eye(2)
    for i in range(n-1):
        left = I - (h/2)*A
        right = (I + (h/2)*A) @ y[i]
        y[i+1] = np.linalg.solve(left, right)
    return t, y[:, 0]

# 5. RK45 из SciPy
def runge_kutta():
    def system(t, y):
        return A @ y
    sol = solve_ivp(system, [0, t_end], y0, method='RK45', 
                    t_eval=np.linspace(0, t_end, int(t_end/h) + 1))
    return sol.t, sol.y[0]

t_exact = np.linspace(0, t_end, 1000)
x_exact = exact_solution(t_exact)

t_exp, x_exp = explicit_euler()
t_imp, x_imp = implicit_euler()
t_semi, x_semi = semi_euler()
t_rk, x_rk = runge_kutta()

plt.figure()

plt.plot(t_exact, x_exact, 'k-', linewidth=2, label='Точное')
plt.plot(t_exp, x_exp, 'r--', label=f'Явный Эйлер (h={h})')
plt.plot(t_imp, x_imp, 'g--', label=f'Неявный Эйлер (h={h})')
plt.plot(t_semi, x_semi, 'b--', label=f'Полуявный (h={h})')
plt.plot(t_rk, x_rk, 'm:', linewidth=2, label='RK45')
plt.xlabel('Время t')
plt.ylabel('x(t)')
plt.title(f'Случай ({case}), m={m}, p={p}, k={k}')
plt.legend()
plt.grid(True)


plt.figure()
plt.plot(t_exp, np.abs(x_exp - exact_solution(t_exp)), 'r-', label='Явный Эйлер')
plt.plot(t_imp, np.abs(x_imp - exact_solution(t_imp)), 'g-', label='Неявный Эйлер')
plt.plot(t_semi, np.abs(x_semi - exact_solution(t_semi)), 'b-', label='Полуявный')
plt.plot(t_rk, np.abs(x_rk - exact_solution(t_rk)), 'm-', label='RK45')
plt.xlabel('Время t')
plt.ylabel('Погрешность')
plt.title('Погрешности методов')
plt.legend()
plt.grid(True)

plt.show()
```

```python
# -*- coding: utf-8 -*-
"""
Численное решение системы уравнений затухающих колебаний
Сравнение различных методов интегрирования
"""

import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
from typing import Tuple

# ======================= НАСТРОЙКИ =======================
TASK_VARIANT = 'a'  # Выбор задачи: 'a' или 'b'
TIME_STEP = 0.1     # Шаг дискретизации
SIMULATION_TIME = 20  # Конечное время моделирования

# ======================= ПАРАМЕТРЫ СИСТЕМЫ =======================
if TASK_VARIANT == 'a':
    mass, damping, stiffness = 1.0, 1.0, 1.0
else:
    mass, damping, stiffness = 1.0, 0.75, 0.125

# Начальные условия [положение, скорость]
initial_state = [1.0, 0.0]

# Матрица системы линейных ДУ
system_matrix = np.array([[0.0, 1.0], 
                          [-stiffness/mass, -damping/mass]])

print(f"Режим: {TASK_VARIANT} | Параметры: m={mass}, p={damping}, k={stiffness}")
print(f"Начальные условия: x(0)={initial_state[0]}, v(0)={initial_state[1]}")
print("-" * 60)

# ======================= ТОЧНОЕ АНАЛИТИЧЕСКОЕ РЕШЕНИЕ =======================
def analytical_solution(time: np.ndarray) -> np.ndarray:
    """Вычисляет точное решение для заданной задачи"""
    if TASK_VARIANT == 'a':
        omega = np.sqrt(3)
        return (np.exp(-time/2) * 
                (np.cos(omega*time/2) + (1/omega)*np.sin(omega*time/2)))
    return 2*np.exp(-time/4) - np.exp(-time/2)

# ======================= ЧИСЛЕННЫЕ МЕТОДЫ =======================
def explicit_euler_method() -> Tuple[np.ndarray, np.ndarray]:
    """Явный метод Эйлера"""
    steps_count = int(SIMULATION_TIME / TIME_STEP) + 1
    time_grid = np.linspace(0, SIMULATION_TIME, steps_count)
    state = np.zeros((steps_count, 2))
    state[0] = initial_state
    
    for step in range(steps_count - 1):
        state[step + 1] = state[step] + TIME_STEP * (system_matrix @ state[step])
    
    return time_grid, state[:, 0]

def implicit_euler_method() -> Tuple[np.ndarray, np.ndarray]:
    """Неявный метод Эйлера"""
    steps_count = int(SIMULATION_TIME / TIME_STEP) + 1
    time_grid = np.linspace(0, SIMULATION_TIME, steps_count)
    state = np.zeros((steps_count, 2))
    state[0] = initial_state
    identity_matrix = np.eye(2)
    
    for step in range(steps_count - 1):
        state[step + 1] = np.linalg.solve(
            identity_matrix - TIME_STEP * system_matrix, 
            state[step]
        )
    
    return time_grid, state[:, 0]

def trapezoidal_method() -> Tuple[np.ndarray, np.ndarray]:
    """Метод трапеций (полуявный Эйлер)"""
    steps_count = int(SIMULATION_TIME / TIME_STEP) + 1
    time_grid = np.linspace(0, SIMULATION_TIME, steps_count)
    state = np.zeros((steps_count, 2))
    state[0] = initial_state
    identity_matrix = np.eye(2)
    half_step = TIME_STEP / 2
    
    for step in range(steps_count - 1):
        left_matrix = identity_matrix - half_step * system_matrix
        right_vector = (identity_matrix + half_step * system_matrix) @ state[step]
        state[step + 1] = np.linalg.solve(left_matrix, right_vector)
    
    return time_grid, state[:, 0]

def runge_kutta_adaptive() -> Tuple[np.ndarray, np.ndarray]:
    """Адаптивный метод Рунге-Кутты 4-5 порядка (RK45)"""
    def dynamics(t: float, y: np.ndarray) -> np.ndarray:
        return system_matrix @ y
    
    solution = solve_ivp(
        dynamics, [0, SIMULATION_TIME], initial_state, 
        method='RK45',
        t_eval=np.linspace(0, SIMULATION_TIME, int(SIMULATION_TIME/TIME_STEP) + 1)
    )
    return solution.t, solution.y[0]

# ======================= ВЫЧИСЛЕНИЕ РЕШЕНИЙ =======================
print("Вычисление решений...")
fine_time_grid = np.linspace(0, SIMULATION_TIME, 1000)
exact_solution = analytical_solution(fine_time_grid)

time_explicit, solution_explicit = explicit_euler_method()
time_implicit, solution_implicit = implicit_euler_method()
time_trapezoidal, solution_trapezoidal = trapezoidal_method()
time_rk, solution_rk = runge_kutta_adaptive()

# ======================= ВИЗУАЛИЗАЦИЯ =======================
plt.style.use('default')
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 10))

# График решений
ax1.plot(fine_time_grid, exact_solution, 'k-', linewidth=3, label='Аналитическое решение')
ax1.plot(time_explicit, solution_explicit, 'r--', linewidth=2, 
         label=f'Явный Эйлер (h={TIME_STEP})')
ax1.plot(time_implicit, solution_implicit, 'g--', linewidth=2, 
         label=f'Неявный Эйлер (h={TIME_STEP})')
ax1.plot(time_trapezoidal, solution_trapezoidal, 'b--', linewidth=2, 
         label=f'Метод трапеций (h={TIME_STEP})')
ax1.plot(time_rk, solution_rk, 'm:', linewidth=3, 
         label='RK45 (адаптивный)')

ax1.set_xlabel('Время t')
ax1.set_ylabel('Перемещение x(t)')
ax1.set_title(f'Сравнение методов интегрирования\nЗадача {TASK_VARIANT}: m={mass}, p={damping}, k={stiffness}', 
              fontsize=14, fontweight='bold')
ax1.legend()
ax1.grid(True, alpha=0.3)
ax1.set_xlim(0, SIMULATION_TIME)

# График погрешностей
ax2.semilogy(time_explicit, np.abs(solution_explicit - analytical_solution(time_explicit)), 
             'r-', linewidth=2, label='Явный Эйлер')
ax2.semilogy(time_implicit, np.abs(solution_implicit - analytical_solution(time_implicit)), 
             'g-', linewidth=2, label='Неявный Эйлер')
ax2.semilogy(time_trapezoidal, np.abs(solution_trapezoidal - analytical_solution(time_trapezoidal)), 
             'b-', linewidth=2, label='Метод трапеций')
ax2.semilogy(time_rk, np.abs(solution_rk - analytical_solution(time_rk)), 
             'm-', linewidth=2, label='RK45')

ax2.set_xlabel('Время t')
ax2.set_ylabel('Абсолютная погрешность |Δx|')
ax2.set_title('Анализ точности методов', fontsize=14, fontweight='bold')
ax2.legend()
ax2.grid(True, alpha=0.3)
ax2.set_xlim(0, SIMULATION_TIME)

plt.tight_layout()
plt.show()

print("✓ Графики построены успешно!")
```

```python
# -*- coding: utf-8 -*-
"""
Численное решение системы уравнений затухающих колебаний
Исследование погрешности при шаге h = 0.001 (вариант b)
"""

import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
from typing import Tuple

TASK_VARIANT = 'b'
TIME_STEP = 0.001          # уменьшенный шаг
SIMULATION_TIME = 20

# параметры варианта b
mass, damping, stiffness = 1.0, 0.75, 0.125
initial_state = [1.0, 0.0]

system_matrix = np.array([[0.0, 1.0],
                          [-stiffness/mass, -damping/mass]])

# точное решение для случая b
def analytical_solution(time: np.ndarray) -> np.ndarray:
    return 2*np.exp(-time/4) - np.exp(-time/2)

# ======================= ЧИСЛЕННЫЕ МЕТОДЫ =======================
def explicit_euler_method() -> Tuple[np.ndarray, np.ndarray]:
    n = int(SIMULATION_TIME / TIME_STEP) + 1
    t = np.linspace(0, SIMULATION_TIME, n)
    y = np.zeros((n, 2))
    y[0] = initial_state
    for i in range(n - 1):
        y[i + 1] = y[i] + TIME_STEP * (system_matrix @ y[i])
    return t, y[:, 0]

def implicit_euler_method() -> Tuple[np.ndarray, np.ndarray]:
    n = int(SIMULATION_TIME / TIME_STEP) + 1
    t = np.linspace(0, SIMULATION_TIME, n)
    y = np.zeros((n, 2))
    y[0] = initial_state
    I = np.eye(2)
    for i in range(n - 1):
        y[i + 1] = np.linalg.solve(I - TIME_STEP * system_matrix, y[i])
    return t, y[:, 0]

def trapezoidal_method() -> Tuple[np.ndarray, np.ndarray]:
    n = int(SIMULATION_TIME / TIME_STEP) + 1
    t = np.linspace(0, SIMULATION_TIME, n)
    y = np.zeros((n, 2))
    y[0] = initial_state
    I = np.eye(2)
    h2 = TIME_STEP / 2
    for i in range(n - 1):
        left = I - h2 * system_matrix
        right = (I + h2 * system_matrix) @ y[i]
        y[i + 1] = np.linalg.solve(left, right)
    return t, y[:, 0]

def runge_kutta_adaptive() -> Tuple[np.ndarray, np.ndarray]:
    def system(t, y):
        return system_matrix @ y
    sol = solve_ivp(
        system, [0, SIMULATION_TIME], initial_state,
        method='RK45',
        t_eval=np.linspace(0, SIMULATION_TIME,
                           int(SIMULATION_TIME / TIME_STEP) + 1)
    )
    return sol.t, sol.y[0]

# ======================= РЕШЕНИЯ =======================
t_fine = np.linspace(0, SIMULATION_TIME, 1000)
x_exact = analytical_solution(t_fine)

t_exp, x_exp = explicit_euler_method()
t_imp, x_imp = implicit_euler_method()
t_tr,  x_tr  = trapezoidal_method()
t_rk,  x_rk  = runge_kutta_adaptive()

# ===== 1‑й график: решения =====
plt.figure(figsize=(8, 6))
plt.plot(t_fine, x_exact, 'k-', linewidth=2.5, label='Аналитическое решение')
plt.plot(t_exp, x_exp, 'r--', label=f'Явный Эйлер (h={TIME_STEP})')
plt.plot(t_imp, x_imp, 'g--', label=f'Неявный Эйлер (h={TIME_STEP})')
plt.plot(t_tr,  x_tr,  'b--', label=f'Метод трапеций (h={TIME_STEP})')
plt.plot(t_rk,  x_rk,  'm:', linewidth=2, label='RK45')
plt.xlabel('Время t')
plt.ylabel('Перемещение x(t)')
plt.title(f'Решение ДУ при шаге h={TIME_STEP} (вариант b)')
plt.legend()
plt.grid(True, alpha=0.3)

# ===== 2‑й график: погрешности (как раньше) =====
plt.figure(figsize=(8, 6))
plt.plot(t_exp, np.abs(x_exp - analytical_solution(t_exp)), 'r-', label='Явный Эйлер')
plt.plot(t_imp, np.abs(x_imp - analytical_solution(t_imp)), 'g-', label='Неявный Эйлер')
plt.plot(t_tr,  np.abs(x_tr  - analytical_solution(t_tr)),  'b-', label='Метод трапеций')
plt.plot(t_rk,  np.abs(x_rk  - analytical_solution(t_rk)),  'm-', label='RK45')
plt.xlabel('Время t')
plt.ylabel('Погрешность')
plt.title(f'Погрешности методов при шаге h={TIME_STEP} (вариант b)')
plt.legend()
plt.grid(True)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Параметры для вашего отрезка [-π, 0] и функции cos(x)
a = -np.pi
b = 0
n = 1000
x = np.linspace(a, b, n)
y = np.cos(x)  # Исследуемая функция f(x) = cos(x)

m = 10  # Максимальное число узлов

XPN = [x[0]]
YPN = [y[0]]

dch = np.zeros(m)
dp = np.zeros(m)

for i in range(1, m + 1):
    # Чебышевские узлы
    t = np.cos((2 * np.arange(1, i + 1) - 1) * np.pi / (2 * i))
    XchN = (a + b) / 2 + (b - a) * t / 2
    YchN = np.cos(XchN)

    # Полином по Чебышевским узлам
    W = np.vander(XchN, increasing=True)  # increasing=True для стабильности
    A = np.linalg.solve(W, YchN)
    Pch = np.polyval(A[::-1], x)  # Обратный порядок коэффициентов

    # Равномерная сетка
    W = np.vander(XPN, increasing=True)
    A = np.linalg.solve(W, YPN)
    Pp = np.polyval(A[::-1], x)

    # Обновление равномерной сетки
    hx = (b - a) / i
    XPN = [a + j * hx for j in range(i + 1)]
    YPN = np.cos(np.array(XPN))

    # График ошибок интерполяции
    plt.figure(figsize=(10, 4))
    
    plt.subplot(1, 2, 1)
    plt.plot(x, np.abs(y - Pp), 'b-', linewidth=2, label='Равномерная сетка')
    plt.plot(x, np.abs(y - Pch), 'r-', linewidth=2, label='Чебышевские узлы')
    plt.xlabel('x')
    plt.ylabel('|Ошибка|')
    plt.title(f'Ошибки интерполяции, i={i}')
    plt.legend()
    plt.grid(True)
    plt.yscale('log')
    
    # Маркеры узлов
    plt.plot(XPN, np.zeros_like(XPN), 'bo', markersize=4, alpha=0.7)
    plt.plot(XchN, np.zeros_like(XchN), 'ro', markersize=4, alpha=0.7)
    
    plt.subplot(1, 2, 2)
    plt.plot(x, y, 'k-', linewidth=1, alpha=0.5, label='cos(x)')
    plt.plot(x, Pch, 'r--', linewidth=1.5, label='Чебыш. интерпол.')
    plt.plot(x, Pp, 'b--', linewidth=1.5, label='Равном. интерпол.')
    plt.plot(XchN, np.cos(XchN), 'ro', markersize=6, label='Чебыш. узлы')
    plt.plot(XPN, np.cos(np.array(XPN)), 'bo', markersize=6, label='Равном. узлы')
    plt.xlabel('x')
    plt.ylabel('f(x)')
    plt.title(f'Интерполяция cos(x), i={i}')
    plt.legend()
    plt.grid(True)
    
    plt.tight_layout()
    plt.show()

    # Максимальные ошибки
    dch[i - 1] = np.max(np.abs(y - Pch))
    dp[i - 1] = np.max(np.abs(y - Pp))

# График логарифма максимальных ошибок
plt.figure(figsize=(8, 6))
plt.semilogy(range(1, m + 1), dch, 'ro-', linewidth=2, markersize=8, label='Чебышевские узлы')
plt.semilogy(range(1, m + 1), dp, 'bo-', linewidth=2, markersize=8, label='Равномерная сетка')
plt.xlabel('Степень полинома n')
plt.ylabel('log₁₀(максимальная ошибка)')
plt.title('Сходимость интерполяции для cos(x) на [-π, 0]')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# Вывод результатов
print("Максимальные ошибки:")
print("Чебыш.:", [f"{e:.2e}" for e in dch])
print("Равном.:", [f"{e:.2e}" for e in dp])
print(f"Соотношение ошибок (равном./чебыш.): {np.max(dp/dch):.1f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

def rate_inaccuracy(a, b, m, add_inaccuracy):
    k = 1000
    h = (b - a) / (k - 1)

    x = np.arange(a, b + h, h)
    y = np.log(x) * np.sin(x)

    pogr = np.zeros(m + 1)

    for i in range(1, m + 1):
        hx = (b - a) / (i + 1)
        XN = np.arange(a, b + hx / 2, hx)
        if add_inaccuracy:
            YN = np.log(XN) * np.sin(XN) + 0.01 * (np.random.rand(len(XN)) - 0.5)
        else:
            YN = np.log(XN) * np.sin(XN)

        W = np.vander(XN)
        A = np.linalg.solve(W, YN)
        P = np.polyval(A, x)

        pogr[i] = np.max(np.abs(y - P))

        plt.plot(x, abs(y - P))
        title = f"Степень = {i}. Границы - [{a}, {b}]." + ("\nСлучайная погрешность внесена" if add_inaccuracy else "")
        plt.title(title)
        plt.show()

    plt.semilogy(pogr[1:16], color="red")
    xticks = plt.xticks()[0]
    plt.xticks(xticks, xticks + 1)
    title = (f"Поведение погрешности при разных степенях. m = {m}"
             + ("\nСлучайная погрешность внесена" if add_inaccuracy else ""))
    plt.title(title)
    plt.show()

'''
Для пункта 1:
a = [1, 1]
b = [np.pi, 6 * np.pi]
m = 15
inaccuracy = False

Для пункта 2:
a = 1
b = 2.12762
m = 30
inaccuracy = False

Для пункта 3:
a = [1, 1, 1]
b = [2.12762, np.pi, 6 * np.pi]
m = [15, 30]
inaccuracy = True
'''

rate_inaccuracy(1, np.pi, 15, False)
```
